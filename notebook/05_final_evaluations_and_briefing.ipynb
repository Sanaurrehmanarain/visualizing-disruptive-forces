```python
import matplotlib.pyplot as plt
from pathlib import Path

FIGURES_DIR = Path("../figures")
FIGURES_DIR.mkdir(parents=True, exist_ok=True)

# Non-technical visual: The "1% Problem"
# Communicating asymmetric risk to a policymaker without network jargon.
labels = ['Top 1% of Cross-Border\nLending Connections', 'Remaining 99% of\nLending Connections']
values = [34.3, 65.7] # Percentages calculated from Member B's analysis
colors = ['#c0392b', '#bdc3c7']

plt.figure(figsize=(9, 4), dpi=300)
bars = plt.barh(labels, values, color=colors, height=0.5)

# Add text labels inside the bars
for bar in bars:
    width = bar.get_width()
    plt.text(width - 3, bar.get_y() + bar.get_height()/2, f'{width}%', 
             ha='right', va='center', color='white', fontweight='bold', fontsize=14)

plt.title('The Systemic Bottleneck:\nExtreme Concentration of Global Financial Risk (2008)', 
          fontweight='bold', fontsize=15, pad=15)
plt.xlabel('Percentage of Total Global Exposure', fontweight='bold', fontsize=12)
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.gca().invert_yaxis() # Put the 1% bar on top
plt.tight_layout()
plt.savefig(FIGURES_DIR / "fig_step4_briefing_visual.png")
plt.show()

print("--- Briefing Visualization Generated ---")
```
